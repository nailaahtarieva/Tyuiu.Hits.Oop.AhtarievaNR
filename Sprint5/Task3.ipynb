### Вариант задания : 3


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Employee в C#, который будет представлять информацию о
сотрудниках компании. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу Employee:
• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).

• Методы:

    o GetInfo(): метод для получения информации о сотруднике в виде строки.

    o Work(): метод для вывода сообщения о выполнении рабочих обязанностей.

    o TakeVacation(): метод для вывода сообщения о взятии отпуска.

Требования к производным классам:
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть
переопределен для добавления информации о управлении командой.
2. Разработчик (Developer): Должен содержать дополнительные атрибуты,
такие как Специализация (Specialization). Метод TakeVacation() должен быть
переопределен для добавления информации о необходимости согласования
отпуска с проектными сроками.
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle).
Метод Work() должен быть переопределен для добавления информации о
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) используйтие в проекте коллекции, делегаты, события.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void WorkDelegate();

public interface IProjectWork
{
    void StartProject(string projectNameStart);
    void CompleteProject(string projectNameEnd);
}

public class Employee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public decimal Salary { get; set; }
    public string Department { get; set; }
    public string Email { get; set; }
    
    public Employee(string name, int age, decimal salary, string department, string email)
    {
        Name = name; 
        Age = age; 
        Salary = salary; 
        Department = department;
        Email = email; 
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Возраст: {Age}, Зарплата {Salary}");
    }

    public virtual void Work()
    {
        Console.WriteLine($"Сотрудник {Name} выполняет свои рабочие обязанности");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"Сотрудник {Name} берет отпуск");
    }

    public virtual void DepartmentMeeting()
    {
        Console.WriteLine($"Сотрудник {Name} участвует в собрании отдела {Department}");
    }
}

List<Employee> employeeList = new List<Employee>();

public class Manager: Employee
{
    private int SubCount{ get; set; }
    private string HolydayStart { get; set; }
    private string HolydayEnd { get; set; }

    public Manager(string name, int age, decimal salary, int subCount, string start, string end, string department, string email): base(name, age, salary, department, email)
    {
        SubCount = subCount; 
        HolydayStart = start; 
        HolydayEnd = end; 
    }

    public override void Work() 
    {
        Console.WriteLine($"Менеджер {Name} управляет командой из {SubCount} сотрудников");
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Возраст: {Age}, Зарплата {Salary}, Подчиненных: {SubCount}");
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Менеджер {Name} берет отпуск с {HolydayStart} по {HolydayEnd}");
    }
}

public class Developer : Employee, IProjectWork
{
    private string Special{ get; set; }
    private string Language { get; set; }

    public Developer(string name, int age, decimal salary, string special, string department, string email, string language): base(name, age, salary, department, email)
    {
        Special = special; 
        Language = language;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Возраст: {Age}, Зарплата {Salary} Специализация: {Special}, Язык программирования: {Language}");
    }

    public virtual void WriteCode()
    {
        Console.WriteLine($"Разработчик {Name} пишет код на {Language}");
    }

    public void StartProject(string projectNameStart)
    {
        Console.WriteLine($"Разработчик {Name} начал работу над проектом: {projectNameStart}");
    }

    public void CompleteProject(string projectNameEnd)
    {
        Console.WriteLine($"Разработчик {Name} завершил проект: {projectNameEnd}");
    }
}

public class Designer: Employee, IProjectWork
{
    private string Style{ get; set; }
    private string DesignProgramms { get; set; }

    public Designer(string name, int age, decimal salary, string department, string email, string style, string programms): base(name, age, salary, department, email)
    {
        Style = style; 
        DesignProgramms = programms;
    }

    public override void Work()
    {
        Console.WriteLine($"Дизайнер {Name} в работе над дизайн-проектами использует стиль {Style} и программу {DesignProgramms}");
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Возраст: {Age}, Зарплата {Salary}, Стиль дизайна: {Style}, используемые программы: {DesignProgramms}");
    }

    public void StartProject(string projectNameStart)
    {
        Console.WriteLine($"Дизайнер {Name} начал работу над проектом: {projectNameStart}");
    }

    public void CompleteProject(string projectNameEnd)
    {
        Console.WriteLine($"Дизайнер {Name} завершил проект: {projectNameEnd}");
    }
}

public class JuniorDeveloper: Developer
{
    private string Mentor { get; set; }

    public JuniorDeveloper(string name, int age, decimal salary, string special, string department, string email, string language, string mentor): base(name, age, salary, special, department, email, language)
    {
        Mentor = mentor;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}, Возраст: {Age}, Зарплата {Salary}, Ментор: {Mentor}");
    }

    public override void WriteCode()
    {
        Console.WriteLine($"Младший разработчик {Name} пишет код под руководством {Mentor}");
    }
}

Manager manager = new Manager("Анна Иванова", 35, 100000, 3, "15 марта", "29 марта", "Управление", "annaivanova@company.com");
Developer developer = new Developer("Петр Кузнецов", 32, 90000, "FullStack", "Разработка", "peterkuznezov@company.com", "C#/JavaScript");
Designer designer = new Designer("Мария Васильева", 28, 70000, "Дизайн", "mariavasilieva@company.com", "Минимализм", "Figma");
JuniorDeveloper juniordev = new JuniorDeveloper("Алексей Смирнов", 20, 50000, "Backend", "Разработка", "alexeysmirnov@company.com", "C#", "Петр Кузнецов");

employeeList.Add(manager);
employeeList.Add(developer);
employeeList.Add(designer);
employeeList.Add(juniordev);

Dictionary<string, Employee> employeeDictionary = new Dictionary<string, Employee>();
employeeDictionary.Add("MAN1", manager);
employeeDictionary.Add("DEV1", developer);
employeeDictionary.Add("DES1", designer);
employeeDictionary.Add("JDEV1", juniordev);

Console.WriteLine("Все сотрудники из List:");
Console.WriteLine("--------------------------");
foreach (var employee in employeeList)
{
    employee.GetInfo();
}

Console.WriteLine("\nДемонстрация делегатов:");
Console.WriteLine("-------------------------");


WorkDelegate workManager = manager.Work;
WorkDelegate workDeveloper = developer.Work;
WorkDelegate workDesigner = designer.Work;

workManager();
workDeveloper();
workDesigner();

Console.WriteLine("\nДемонстрация лямбда-выражений:");
Console.WriteLine("--------------------------------");

var highSalary = employeeList.Where(e => e.Salary > 60000);
Console.WriteLine("Сотрудники с зарплатой выше 60000:");
foreach (var emp in highSalary)
{
    emp.GetInfo();
}

Console.WriteLine("\nДемонстрация работы со словарем:");
Console.WriteLine("---------------------");

if (employeeDictionary.TryGetValue("DEV1", out Employee foundEmployee))
{
    Console.WriteLine("Найденный сотрудник:");
    foundEmployee.GetInfo();
}

Console.WriteLine("\nРабота сотрудников:");
Console.WriteLine("---------------------");

manager.DepartmentMeeting();
manager.TakeVacation();
        
developer.DepartmentMeeting();
developer.WriteCode();
developer.StartProject("Разработка сайта");
developer.CompleteProject("Телеграмм-бот");

designer.DepartmentMeeting();
designer.Work();
designer.StartProject("Дизайн профиля");
designer.CompleteProject("Главная страница");

juniordev.DepartmentMeeting();
juniordev.WriteCode();
```
