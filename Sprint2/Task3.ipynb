### Вариант задания : 3


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Создать базовый класс Employee в C#, который будет представлять информацию о
сотрудниках компании. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу Employee:
• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).

• Методы:

    o GetInfo(): метод для получения информации о сотруднике в виде строки.

    o Work(): метод для вывода сообщения о выполнении рабочих обязанностей.

    o TakeVacation(): метод для вывода сообщения о взятии отпуска.

Требования к производным классам:
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть
переопределен для добавления информации о управлении командой.
2. Разработчик (Developer): Должен содержать дополнительные атрибуты,
такие как Специализация (Specialization). Метод TakeVacation() должен быть
переопределен для добавления информации о необходимости согласования
отпуска с проектными сроками.
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle).
Метод Work() должен быть переопределен для добавления информации о
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к существующим классам (базовому и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public interface IProjectWork
{
    void StartProject(string projectNameStart); // начатый проект
    void CompleteProject(string projectNameEnd); // завершенный проект
}

public class Employee
{
    public string Name { get; set; } // имя
    public int Age { get; set; } // возраст
    public decimal Salary { get; set; } // зарплата
    public string Department { get; set; }  // новый атрибут - отдел
    public string Email { get; set; } // новый атрибут - email
    
    public Employee(string name, int age, decimal salary, string department, string email)
    {
        Name = name; 
        Age = age; 
        Salary = salary; 
        Department = department;
        Email = email; 
    }

    public virtual string GetInfo() // метод для получения информации о сотруднике в виде строки
    {
        return $"Сотрудник: {Name}, Возраст: {Age}, Зарплата {Salary}";
    }

    public virtual void Work() // метод для вывода сообщения о выполнении рабочих обязанностей
    {
        Console.WriteLine($"Сотрудник {Name} выполняет свои рабочие обязанности");
    }

    public virtual void TakeVacation() // метод для вывода сообщения о взятии отпуска
    {
        Console.WriteLine($"Сотрудник {Name} берет отпуск");
    }

    public virtual void DepartmentMeeting() // новый метод для получения информации о том, к какому отделу принадлежит сотрудник
    {
        Console.WriteLine($"Сотрудник {Name} участвует в собрании отдела {Department}");
    }
    
}

public class Manager: Employee
{
    private int SubCount{ get; set; } // количество подчиненных
    private string HolydayStart { get; set; } // дата выхода в отпуск
    private string HolydayEnd { get; set; } // дата выхода с отпуска

    public Manager(string name, int age, decimal salary, int subCount, string start, string end, string department, string email): base(name, age, salary, department, email)
    {
        SubCount = subCount; 
        HolydayStart = start; 
        HolydayEnd = end; 
    }

    public override void Work()
    {
        Console.WriteLine($"Менеджер {Name} управляет командой из {SubCount} сотрудников");
    }

    public override string GetInfo()
    {
            return $"{base.GetInfo()}, Подчиненных: {SubCount}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Менеджер {Name} берет отпуск с {HolydayStart} по {HolydayEnd}");
    }
}

public class Developer : Employee, IProjectWork
{
    private string Special{ get; set; } // специализация
    private string Language { get; set; } // новый атрибут - язык программирования
    private string ProjectNameStart { get; set; } // новый атрибут для интерфейса - начатый проект
    private string ProjectNameEnd { get; set; } // новый атрибут для интерфейса - законченный проект

    public Developer(string name, int age, decimal salary, string special, string department, string email, string language, string pstart, string pend): base(name, age, salary, department, email)
    {
        Special = special; 
        Language = language;
        ProjectNameStart = pstart;
        ProjectNameEnd = pend;
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Разработчик {Name}, согласует отпуск с проектными сроками");
        Console.WriteLine("Срок сдачи проекта: 1 марта");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Специализация: {Special}, Язык программирования: {Language}";
    }

    public virtual void WriteCode() // новый метод для вывод сообщения о написанном коде
    {
        Console.WriteLine($"Разработчик {Name} пишет код на {Language}");
    }

    // Реализация интерфейса IProjectWork
    public void StartProject(string pstart)
    {
        Console.WriteLine($"Разработчик {Name} начал работу над проектом: {ProjectNameStart}");
    }

    public void CompleteProject(string pend)
    {
        Console.WriteLine($"Разработчик {Name} завершил проект: {ProjectNameEnd}");
    }
}

public class Designer: Employee, IProjectWork
{
    private string Style{ get; set; } // стиль дизайна
    private string HolydayStart { get; set; } // дата выхода в отпуск
    private string HolydayEnd { get; set; } // дата выхода  с отпуска
    private string DesignProgramms { get; set; } // новый атрибут - используемые программы
    private string ProjectNameStart { get; set; } // новый атрибут для интерфейса - начатый проект
    private string ProjectNameEnd { get; set; } // новый атрибут для интерфейса - законченный проект

    public Designer(string name, int age, decimal salary, string department, string email, string style, string start, string end, string programms, string pstart, string pend): base(name, age, salary, department, email)
    {
        Style = style; 
        HolydayStart = start; 
        HolydayEnd = end; 
        DesignProgramms = programms;
        ProjectNameStart = pstart;
        ProjectNameEnd = pend;
    }

    public override void Work()
    {
        Console.WriteLine($"Дизайнер {Name} в работе над дизайн-проектами использует стиль {Style} и программу {DesignProgramms}");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Стиль дизайна: {Style}, исользуемые программы: {DesignProgramms}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Дизайнер {Name} берет отпуск с {HolydayStart} по {HolydayEnd}");
    }

    // Реализация интерфейса IProjectWork
    public void StartProject(string pstart)
    {
        Console.WriteLine($"Дизайнер {Name} начал работу над проектом: {ProjectNameStart}");
    }

    public void CompleteProject(string pend)
    {
        Console.WriteLine($"Дизайнер {Name} завершил проект: {ProjectNameEnd}");
    }
}

public class JuniorDeveloper: Developer
{
    private string Mentor { get; set; } // наставник

    public JuniorDeveloper(string name, int age, decimal salary, string special, string department, string email, string language, string pstart, string pend, string mentor): base(name, age, salary, special, department, email, language, pstart, pend)
    {
        Mentor = mentor;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Ментор: {Mentor}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Младший разработчик {Name} согласует отпуск со старшим разработчиком {Mentor}");
    }

    public void SubmitReport()
    {
        Console.WriteLine($"Младший разработчик {Name} отчитался старшему разработчику {Mentor} о проделанной работе");
    }

    public void Skills()
    {
        Console.WriteLine($"Младший разработчик {Name} начал изучать JavaScript");
    }

    public override void WriteCode()
    {
        Console.WriteLine($"Младший разработчик {Name} пишет код под руководством {Mentor}");
    }
}

Manager manager = new Manager("Анна Иванова", 35, 100000, 3, "15 марта", "29 марта", "Управление", "annaivanova@company.com");
        
Developer developer = new Developer("Петр Кузнецов", 32, 90000, "FullStack", "Разработка", "peterkuznezov@company.com", "C#/JavaScript", "Разработка сайта", "Разработка телеграмм-бота");
        
Designer designer = new Designer("Мария Васильева", 28, 70000, "Дизайн", "mariavasilieva@company.com", "Минимализм", "4 октября", "18 октября", "Figma", "Дизайн профиля пользователя", "Дизайн главной страницы");
        
JuniorDeveloper juniordev = new JuniorDeveloper("Алексей Смирнов", 20, 50000, "Backend", "Разработка", "alexeysmirnov@company.com", "C#", "Изучение основ", "Первый учебный проект", "Петр Кузнецов");

Console.WriteLine("\n" + manager.GetInfo());
manager.Work();
manager.DepartmentMeeting();
manager.TakeVacation();
        
Console.WriteLine("\n" + developer.GetInfo());
developer.DepartmentMeeting();
developer.WriteCode();
developer.CompleteProject("Разработка телеграмм-бота");
developer.StartProject("Разработка сайта");
developer.TakeVacation();

Console.WriteLine("\n" + designer.GetInfo());
designer.DepartmentMeeting();
designer.Work();
designer.CompleteProject("Дизайн главной страницы");
designer.StartProject("Дизайн профиля пользователя");
designer.TakeVacation();

Console.WriteLine("\n" + juniordev.GetInfo());
juniordev.Skills();
juniordev.WriteCode();
juniordev.SubmitReport();
juniordev.TakeVacation();
```
