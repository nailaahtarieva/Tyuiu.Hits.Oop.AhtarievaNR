### Вариант задания : 3


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Создать базовый класс Employee в C#, который будет представлять информацию о
сотрудниках компании. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу Employee:
• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).

• Методы:

    o GetInfo(): метод для получения информации о сотруднике в виде строки.

    o Work(): метод для вывода сообщения о выполнении рабочих обязанностей.

    o TakeVacation(): метод для вывода сообщения о взятии отпуска.

Требования к производным классам:
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть
переопределен для добавления информации о управлении командой.
2. Разработчик (Developer): Должен содержать дополнительные атрибуты,
такие как Специализация (Specialization). Метод TakeVacation() должен быть
переопределен для добавления информации о необходимости согласования
отпуска с проектными сроками.
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle).
Метод Work() должен быть переопределен для добавления информации о
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к существующим классам (базовому и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

public class Employee // базовый класс 
{
    public string Name { get; set; } 
    public int Age { get; set; }
    public decimal Salary { get; set; }
    public string Department { get; set; } // новый атрибут - отдел
    public DateTime HireDate { get; set; } // новый атрибут - дата приема на работу
    public string Email { get; set; } // новый атрибут - email
    public int YearsOfExperience { get; set; } // новый атрибут - опыт работы
    
    public Employee(string name, int age, decimal salary, string department, DateTime hireDate, string email, int experience)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Department = department;
        HireDate = hireDate;
        Email = email;
        YearsOfExperience = experience;
    }

    public virtual string GetInfo()
    {
        return $"Сотрудник: {Name}, Возраст: {Age}, Зарплата: {Salary}, Отдел: {Department}, Опыт работы: {YearsOfExperience} лет";
    }

    public virtual void Work()
    {
        Console.WriteLine($"Сотрудник {Name} выполняет свои рабочие обязанности");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"Сотрудник {Name} берет отпуск");
    }

    // Новые методы
    public virtual void AttendMeeting()
    {
        Console.WriteLine($"Сотрудник {Name} участвует в собрании отдела {Department}");
    }
    
    public virtual void SubmitReport()
    {
        Console.WriteLine($"Сотрудник {Name} сдает отчет по работе");
    }
    
    public int CalculateYearsInCompany()
    {
        return DateTime.Now.Year - HireDate.Year;
    }
    
}

public interface IProjectWork
{
    void StartProject(string projectName); // начатый проект
    void CompleteProject(string projectName); // завершенный проект
    int GetActiveProjectsCount(); // число активных проектов
}

public interface IManagement
{
    void HireEmployee(string employeeName); // наем сотрудника
    void ConductPerformanceReview(string employeeName); // проверка эффективности
}

public class Manager : Employee, IProjectWork, IManagement
{
    private int SubCount { get; set; }
    private string HolydayStart { get; set; }
    private string HolydayEnd { get; set; }
    private List<string> ActiveProjects { get; set; } // новый атрибут - активные проекты
    private decimal Budget { get; set; } // новый атрибут - бюджет

    public Manager(string name, int age, decimal salary, string department, DateTime hireDate, string email, int experience, int subCount, string start, string end, decimal budget) : base(name, age, salary, department, hireDate, email, experience)
    {
        SubCount = subCount;
        HolydayStart = start;
        HolydayEnd = end;
        ActiveProjects = new List<string>();
        Budget = budget;
    }

    public override void Work()
    {
        Console.WriteLine($"Менеджер {Name} управляет командой из {SubCount} сотрудников с бюджетом {Budget:C}");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Подчиненных: {SubCount}, Бюджет: {Budget}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Менеджер {Name} берет отпуск с {HolydayStart} по {HolydayEnd}");
    }

    // Реализация интерфейса IProjectWork
    public void StartProject(string projectName) // реализация наследуемого метода о начатом проекте
    {
        ActiveProjects.Add(projectName);
        Console.WriteLine($"Менеджер {Name} запустил проект: {projectName}");
    }

    public void CompleteProject(string projectName)  // реализация наследуемого метода о завершенном проекте
    {
        Console.WriteLine($"Менеджер {Name} завершил проект: {projectName}");
    }

    public int GetActiveProjectsCount() // число активных проектов
    {
        return ActiveProjects.Count;
    }

    // Реализация интерфейса IManagement
    public void HireEmployee(string employeeName) // наем нового сотрудника
    {
        Console.WriteLine($"Менеджер {Name} нанял нового сотрудника: {employeeName}");
    }

    public void ConductPerformanceReview(string employeeName) // оценка эффективности проекта
    {
        Console.WriteLine($"Менеджер {Name} проводит оценку эффективности для: {employeeName}");
    }

    public void ConductTeamMeeting()
    {
        Console.WriteLine($"Менеджер {Name} проводит собрание команды из {SubCount} человек");
    }
}

public class Developer : Employee, IProjectWork
{
    private string Special { get; set; }
    private string ProgrammingLanguage { get; set; } // новый атрибут
    private List<string> CurrentProjects { get; set; } // новый атрибут
    private int CodeCommits { get; set; } // новый атрибут

    public Developer(string name, int age, decimal salary, string department, DateTime hireDate, string email, int experience, string special, string language) : base(name, age, salary, department, hireDate, email, experience)
    {
        Special = special;
        ProgrammingLanguage = language;
        CurrentProjects = new List<string>();
        CodeCommits = 0;
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Разработчик {Name}, специализирующийся на: {Special}, согласует отпуск с проектными сроками");
        Console.WriteLine("Срок сдачи проекта: 1 марта");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Специализация: {Special}, Язык: {ProgrammingLanguage}, Коммиты: {CodeCommits}";
    }
    
    public void WriteCode()
    {
        CodeCommits++;
        Console.WriteLine($"Разработчик {Name} написал код на {ProgrammingLanguage}. Всего коммитов: {CodeCommits}");
    }
    
    // Реализация интерфейса IProjectWork
    public void StartProject(string projectName)
    {
        CurrentProjects.Add(projectName);
        Console.WriteLine($"Разработчик {Name} начал работу над проектом: {projectName}");
    }

    public void CompleteProject(string projectName)
    {
        Console.WriteLine($"Разработчик {Name} завершил проект: {projectName}");
    }

    public int GetActiveProjectsCount()
    {
        return CurrentProjects.Count;
    }

}

public class Designer : Employee, IProjectWork
{
    private string Style { get; set; }
    private string HolydayStart { get; set; }
    private string HolydayEnd { get; set; }
    private string DesignProgramms { get; set; } // новый атрибут - используемые программы
    private int DesignsCreated { get; set; } // новый атрибут - созданные дизайны
    private List<string> DesignProjects { get; set; } // новый атрибут

    public Designer(string name, int age, decimal salary, string department, DateTime hireDate, string email, int experience, string style, string start, string end, string software) : base(name, age, salary, department, hireDate, email, experience)
    {
        Style = style;
        HolydayStart = start;
        HolydayEnd = end;
        DesignProgramms = software;
        DesignsCreated = 0;
        DesignProjects = new List<string>();
    }

    public override void Work()
    {
        Console.WriteLine($"Дизайнер {Name} в работе над дизайн-проектами использует стиль {Style} и программу {DesignProgramms}");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Стиль дизайна: {Style}, Программа: {DesignProgramms}, Создано дизайнов: {DesignsCreated}";
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"Дизайнер {Name} берет отпуск с {HolydayStart} по {HolydayEnd}");
    }

    // Реализация интерфейса IProjectWork
    public void StartProject(string projectName)
    {
        DesignProjects.Add(projectName);
        Console.WriteLine($"Дизайнер {Name} начал дизайн-проект: {projectName}");
    }

    public void CompleteProject(string projectName)
    {
        if (DesignProjects.Remove(projectName))
        {
            DesignsCreated++;
            Console.WriteLine($"Дизайнер {Name} завершил дизайн-проект: {projectName}");
        }
    }

    public int GetActiveProjectsCount()
    {
        return DesignProjects.Count;
    }

    public void CreateDesign()
    {
        DesignsCreated++;
        Console.WriteLine($"Дизайнер {Name} создал дизайн в стиле {Style}. Всего создано: {DesignsCreated}");
    }
    
    public void PresentDesign()
    {
        Console.WriteLine($"Дизайнер {Name} представляет дизайн клиенту");
    }
  
}

public class SeniorDeveloper : Developer, IManagement
{
    private bool IsTechnicalLead { get; set; }
    private int JuniorDevelopersMentored { get; set; }

    public SeniorDeveloper(string name, int age, decimal salary, string department, DateTime hireDate, string email, int experience, string special, string language, bool isTechLead) : base(name, age, salary, department, hireDate, email, experience, special, language)
    {
        IsTechnicalLead = isTechLead;
        JuniorDevelopersMentored = 0;
    }

    // Реализация интерфейса IManagement
    public void HireEmployee(string employeeName)
    {
        Console.WriteLine($"Старший разработчик {Name} участвует в найме: {employeeName}");
    }

    public void ConductPerformanceReview(string employeeName)
    {
        Console.WriteLine($"Старший разработчик {Name} проводит техническое собеседование для: {employeeName}");
    }

    public void AllocateBudget(decimal amount)
    {
        Console.WriteLine($"Старший разработчик {Name} запрашивает бюджет {amount:C} на техническое оборудование");
    }

    public void MentorJuniorDeveloper()
    {
        JuniorDevelopersMentored++;
        Console.WriteLine($"Старший разработчик {Name} обучает младшего разработчика. Всего обучено: {JuniorDevelopersMentored}");
    }
    
    public void ConductCodeReview()
    {
        Console.WriteLine($"Старший разработчик {Name} просматривает код команды");
    }
    
}

Manager manager = new Manager("Анна Иванова", 35, 85000, "Управление", new DateTime(2015, 3, 15), "anna@company.com", 10, 4, "15 марта", "29 марта", 500000);
        
Developer developer = new Developer("Петр Кузнецов", 23, 72000, "Разработка", new DateTime(2020, 6, 10),"peter@company.com", 3, "Backend", "C#");
        
Designer designer = new Designer("Мария Васильева", 32, 65000, "Дизайн", new DateTime(2018, 2, 20),"maria@company.com", 5, "Минимализм", "4 октября", "18 октября", "Figma");
        
SeniorDeveloper seniorDev = new SeniorDeveloper("Алексей Смирнов", 40, 120000, "Разработка", new DateTime(2010, 8, 5), "alexey@company.com", 12, "FullStack", "C#/JavaScript", true);

Console.WriteLine("\n" + manager.GetInfo());
manager.Work();
manager.TakeVacation();
manager.StartProject("Сайт для оказания психологической помощи");
manager.HireEmployee("Иван Петров");
        
Console.WriteLine("\n" + developer.GetInfo());
developer.TakeVacation();
developer.WriteCode();
developer.StartProject("Телеграмм бот");
        
Console.WriteLine("\n" + designer.GetInfo());
designer.Work();
designer.TakeVacation();
designer.CreateDesign();
designer.StartProject("Редизайн главной страницы");

Console.WriteLine("\n" + seniorDev.GetInfo());
seniorDev.Work();
seniorDev.MentorJuniorDeveloper();
seniorDev.ConductCodeReview();
seniorDev.HireEmployee("Ольга Новикова");
        
```
