```polyglot-notebook
using System;

public class Product
{
    private string Name { get; set; }
    private decimal Price { get; set; }
    private int Quantity { get; set; }
    protected string Category { get; set; }
    internal string Description { get; set; }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity = amount;
        Console.WriteLine($"Количество товара на складе уменьшено до: {Quantity}");
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name}, Цена: {Price} руб., Количество: {Quantity}, Категория: {Category}, Описание: {Description}";
    }
}

public class Seller
{
    private string Name { get; set; }
    private string EmployeeId { get; set; }
    protected decimal Salary { get; set; }
    internal string ContactInfo { get; set; }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo; 
    }
    
    public void AddProduct()
    {
        Console.WriteLine($"Добавлен новый продукт в список товаров продавца: Сыр плавленный");
    }

    public void SellProduct()
    {
        Console.WriteLine($"Продавец {Name} продал товар. Количество товара на складе уменьшено.");
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name}, ID: {EmployeeId}, Зарплата: {Salary} руб., Контакты: {ContactInfo}";
    }
}

public class Store 
{
    private string StoreName { get; set; }
    private string Location { get; set; }
    public string StoreHours { get; set; }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;
    }

    public void AddSeller()
    {
        Console.WriteLine($"Добавлен продавец: Иван Иванов");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в магазине: Молоко, Сыр плавленный");
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {StoreName}, Адрес: {Location}, Часы работы: {StoreHours}";
    }
}


Product product = new Product("Молоко", 70, 50, "Молочные продукты", "Молоко коровье");
Seller seller = new Seller("Иван Иванов", "12345", 50000, "+79088754670");
Store store = new Store("Монетка", "Улица народная д 8", "7:00 - 22:30");

       
Console.WriteLine(product.GetProductInfo());
Console.WriteLine($"Общая стоимость товара: {product.GetTotalPrice()} руб.");
product.UpdateQuantity(45);       

Console.WriteLine("\n");
Console.WriteLine(seller.GetSellerInfo());
seller.AddProduct();
seller.SellProduct();

Console.WriteLine("\n");
Console.WriteLine(store.GetStoreInfo());
store.AddSeller();
store.ListProducts();
```
